## Radial quantum number 2 and 3  for both bottomonium and charmonium respectively

## 2S bottomonium Power Law

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.pyplot import figure
```

```python
m_b = 5.174
r_min = 0
r_max = 7
h = 0.0009
h_sq = h**2
r = np.arange(r_min, r_max, h)
U = np.zeros(len(r))
V = np.zeros(len(r))
g_r = np.zeros(len(r))

## trial_energy = np.array([-0.2987])
trial_energy = float(input('Enter the trial energy: '))

for i in range(2, len(r)):
    U[0] = 0 
    U[1] = h
    V[0] = (-8.064+(6.898)*(r_min)**0.1)
    V[1] = (-8.064+(6.898)*(r_min + h)**0.1)
    g_r[0] = -2*(m_b/2)*(V[0]-trial_energy)
    g_r[1] = -2*(m_b/2)*(V[1]-trial_energy)
    
    V[i] =(-8.064+(6.898)*(r[i]**0.1))
    g_r[i] =-2*(m_b/2)*(V[i]-trial_energy)  
    U[i] = (2*(1-(5.0/12)*h_sq*g_r[i-1])*U[i-1]-(1+(1.0/12)*h_sq*g_r[i-2])*U[i-2])/(1+(1.0/12)*h_sq*g_r[i])


plt.plot(r, U/r)
plt.title('E = {}'.format(trial_energy))
plt.axhline(y=0, color='r',linestyle='-')
#plt.ylim(0,1)
plt.xlabel('r')
plt.ylabel('Psi(r)')
#plt.savefig('wavebottom.png')
plt.show()

```

## Normalising the radial wavefunction 

```python
norm_U_2s = np.zeros(len(r))
A = 0
for i in range(0, len(r)):
    A = A + np.abs(U[i])**2*h
for i in range(0, len(r)): 
        norm_U_2s[i] = U[i]/np.sqrt(A)

plt.plot(r, norm_U_2s/r,  label = 'Normalised')
#plt.plot(r, (U/r), label = 'Unnormalised')
plt.axhline(y=0, color='k',linestyle='-')
plt.xlabel('r')
plt.ylabel('PSI_Normalised')
plt.legend() 
plt.show()
```

##  Radial wavefunction at the origin 

```python
R_sq_2s = 1/(A)
print('The radial function at the origin bottomonium is {:.4f} GeV'.format(R_sq_2s))
```

## Mass of the bottomonium quark for 2S
Using this expression $M_{b\bar b} = 2M_{b} + E_{ns} $ where the $E_{ns}$ is the binding energy which has been calculated above

```python
M_bb_2S = 2 * m_b + trial_energy

print('The mass of the bottomonium is {:.4f} GeV'.format(M_bb_2S))
```

## 3S bottomonium Power law

```python
m_b = 5.174
r_min = 0
r_max = 7
h = 0.0009
h_sq = h**2
r = np.arange(r_min, r_max, h)
U = np.zeros(len(r))
V = np.zeros(len(r))
g_r = np.zeros(len(r))

## trial_energy = np.array([0.0367])
trial_energy = float(input('Enter the trial energy: '))

for i in range(2, len(r)):
    U[0] = 0 
    U[1] = h
    V[0] = (-8.064+(6.898)*(r_min)**0.1)
    V[1] = (-8.064+(6.898)*(r_min + h)**0.1)
    g_r[0] = -2*(m_b/2)*(V[0]-trial_energy)
    g_r[1] = -2*(m_b/2)*(V[1]-trial_energy)
    
    V[i] =(-8.064+(6.898)*(r[i]**0.1))
    g_r[i] =-2*(m_b/2)*(V[i]-trial_energy)  
    U[i] = (2*(1-(5.0/12)*h_sq*g_r[i-1])*U[i-1]-(1+(1.0/12)*h_sq*g_r[i-2])*U[i-2])/(1+(1.0/12)*h_sq*g_r[i])


plt.plot(r, U/r)
plt.title('E = {}GeV'.format(trial_energy))
plt.axhline(y=0, color='r',linestyle='-')
#plt.ylim(0,1)
plt.xlabel('r')
plt.ylabel('Psi(r)')
#plt.savefig('wavebottom.png')
plt.show()




```

## Normalising the radial wavefunction at 3S level 
For 3D schrodinger equation problem in a symmetric potential, the normalised reduced radial function is
$$ U(r) = \frac{\phi(r)}{\sqrt{A}}$$ where A is the normalisation integral 

```python
norm_U_3s = np.zeros(len(r))
A = 0
for i in range(0, len(r)):
    A = A + np.abs(U[i])**2*h
for i in range(0, len(r)): 
        norm_U_3s[i] = U[i]/np.sqrt(A)

plt.plot(r, norm_U_3s/r,  label = 'Normalised')
#plt.plot(r, (U/r), label = 'Unnormalised')
plt.axhline(y=0, color='k',linestyle='-')
plt.xlabel('r')
plt.ylabel('PSI_Normalised')
plt.legend() 
plt.show()
```

##  Radial wavefunction at the origin at 3S
The schrodinger wavefunction at the origin is given by  $$ \psi(0) = \frac{U^{'}(0)}{\sqrt{4\pi}}$$
$$\therefore |\psi(0)|^{2} = \frac{1}{4\pi A}$$
Then, $$ |R(0)|^{2} = \frac{1}{A}$$

```python
R_sq_3s = 1/(A)
print('The radial function at the origin bottomonium at 3S level is {:.4f} GeV'.format(R_sq_3s))
```

## Mass of the bottomonium quark for 3S
Using this expression $M_{b\bar b} = 2M_{b} + E_{ns} $ where the $E_{ns}$ is the binding energy which has been calculated above

```python
M_bb_3S = 2 * m_b + trial_energy

print('The mass of the bottomonium at 3S level is {:.4f} GeV'.format(M_bb_3S))
```

## 2S charmonium  power law 

```python
m_c = 1.8
r_min_cc = 0
r_max_cc = 10
h_cc = 0.0009
h_sq_cc = h_cc**2
r_cc = np.arange(r_min_cc, r_max_cc, h_cc)
U_cc = np.zeros(len(r_cc))
V_cc = np.zeros(len(r_cc))
g_r_cc = np.zeros(len(r_cc))

trial_energy_cc = np.array([0.1,0.11,0.102])

#trial_energy_cc = float(input('Enter the trial energy: '))

for j in range(len(trial_energy_cc)):
    for i in range(2, len(r_cc)):
        U_cc[0] = 0 
        U_cc[1] = h_cc
        V_cc[0] = (-8.064+(6.898)*(r_min_cc)**0.1)
        V_cc[1] = (-8.064+(6.898)*(r_min_cc + h_cc)**0.1)
        g_r_cc[0] = -2*(m_c/2)*(V_cc[0]-trial_energy_cc[j])
        g_r_cc[1] = -2*(m_c/2)*(V_cc[1]-trial_energy_cc[j])
    
        V_cc[i] =(-8.064+(6.898)*(r_cc[i]**0.1))
        g_r_cc[i] =-2*(m_c/2)*(V_cc[i]-trial_energy_cc[j])  
        U_cc[i] = (2*(1-(5.0/12)*h_sq_cc*g_r_cc[i-1])*U_cc[i-1]-(1+(1.0/12)*h_sq_cc*g_r_cc[i-2])*U_cc[i-2])/(1+(1.0/12)*h_sq_cc*g_r_cc[i])
        
    
    
    plt.plot(r_cc, (U_cc/r_cc), label = 'Eigen energy {} GeV'.format(trial_energy_cc[j]))
    
plt.title('E = {}GeV'.format(trial_energy_cc))
plt.axhline(y=0, color='r',linestyle='--')
plt.legend()
plt.xlabel('r')
plt.ylabel('Psi(r)')
plt.savefig('wavecharmed.png')
plt.show()
```

## Normalising the radial wavefunction at 2S level

```python
norm_U_cc_2s = np.zeros(len(r_cc))
A_cc = 0
for i in range(0, len(r_cc)):
    A_cc = A_cc + np.abs(U_cc[i])**2*h_cc
for i in range(0, len(r_cc)): 
        norm_U_cc_2s[i] = U_cc[i]/np.sqrt(A_cc)


plt.plot(r_cc, norm_U_cc_2s/r_cc,  label = 'Normalised')
#plt.plot(r, (U/r), label = 'Unnormalised')
plt.axhline(y=0, color='k',linestyle='-')
plt.xlabel('r')
plt.ylabel('PSI_Normalised')
plt.legend()
plt.savefig('wavepowercharmonium.png')
plt.show()
```

##  Radial wavefunction at the origin at 2S

```python
R_sq_cc_2S = 1/(A_cc)
print('The radial function at the origin for charmonium is {:.4f} GeV'.format(R_sq_cc_2S))
```

## Mass of the charmonium quark at 2S level

```python
M_cc_2S = 2 * m_c + trial_energy_cc[2]

print('The mass of the charmonium is {:.4f} GeV'.format(M_cc_2S))
```
