```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.pyplot import figure
```

```python
r_min= 0.1
r_max =5
h_cc = 0.0004
k = 0.52
a = 2.34
r = np.arange(r_min, r_max, h_cc)
V_cornell = np.zeros(len(r))
V_log = np.zeros(len(r))
V_power = np.zeros(len(r))

for i in range(len(r)):
    
    V_cornell[i] = -(k/r[i])+(r[i]/(a**2))
    V_log[i] = -0.6635+(0.733)*np.log(r[i]*1)
    V_power[i] =(-8.064+(6.898)*(r[i]**0.1))
    
plt.title('A plot showing log and power potentials against r')
plt.plot(r,V_log, label = 'Logarithimic Potential')
plt.plot(r, V_power, label = 'Power-law potential')
plt.plot(r, V_cornell, label = 'cornell potential')
plt.xlabel('distance, r')
plt.ylabel('Potentials')
plt.legend()
plt.axhline(y=0, color='r',linestyle='--')


```

wFrom the diamgram above one can see that,  the potentials, independent of heavy quark  flavours, coincide with each other in the region of  the average distances between heavy quarks in the (cc) and 
(bb) systems, so  0.1 fm < r < 1 fm , although those potentials have different asymptotic 
behaviour in the regions of very small (r $\rightarrow 0$ ) and very large (r $\rightarrow \infty$) distances.
