## Bottomonium 

```python
import numpy as np
import matplotlib.pyplot as plt
```

## 2S Bottomonium using Cornell Potential 

```python
k = 0.52
a = 2.34
m_b = 5.18
r_min= 1e-30
r_max =7
h = 0.0009
h_sq = h**2
r = np.arange(r_min, r_max, h)
U = np.zeros(len(r))
V = np.zeros(len(r))
g_r = np.zeros(len(r))

trial_energy = np.array([0.419,0.423,0.4215])
#trial_energy = float(input('Enter the trial energy: '))

for j in range(len(trial_energy)):

    for i in range(2, len(r)):
        U[0] = 0 
        U[1] = h
        V[0] = -(k/r_min)+(r_min/(a**2))
        V[1] = -(k/(r_min+h))+((r_min+h)/(a**2))
        g_r[0] = -2*(m_b/2)*(V[0]-trial_energy[j])
        g_r[1] = -2*(m_b/2)*(V[1]-trial_energy[j])
    
        V[i] = -(k/r[i])+(r[i]/(a**2))
        g_r[i] =-2*(m_b/2)*(V[i]-trial_energy[j])  
        U[i] = (2*(1-(5.0/12)*h_sq*g_r[i-1])*U[i-1]-(1+(1.0/12)*h_sq*g_r[i-2])*U[i-2])/(1+(1.0/12)*h_sq*g_r[i])
    plt.plot(r, (U/r), label = 'Eigen energy {} GeV'.format(trial_energy[j]))


plt.title('E = {} GeV'.format(trial_energy))
plt.axhline(y=0, color='r',linestyle='--')
plt.legend()
plt.xlabel('r')
plt.ylabel('Psi(r)')
plt.savefig('wavecornellbottom2s.png')
plt.show()


```

## Normalising the radial wavefunction at 2S

```python
norm_U = np.zeros(len(r))
A = 0
for i in range(0, len(r)):
    A = A + np.abs(U[i])**2*h
for i in range(0, len(r)): 
        norm_U[i] = U[i]/np.sqrt(A)

plt.plot(r, norm_U/r,  label = 'Normalised')
#plt.plot(r, (U/r), label = 'Unnormalised')
plt.axhline(y=0, color='k',linestyle='-')
plt.xlabel('r')
plt.ylabel('PSI_Normalised')
plt.legend() 
plt.show()
```

##  Radial wavefunction at the origin at 2S

```python
R_sq_2s = 1/(A)
print('The radial function at the origin bottomonium is {:.4f} GeV'.format(R_sq_2s))
```

## Mass of the bottomonium quark for 2S
Using this expression $M_{b\bar b} = 2M_{b} + E_{ns} $ where the $E_{ns}$ is the binding energy which has been calculated above

```python
M_bb_2s = 2 * m_b + trial_energy[2]

print('The mass of the bottomonium is {:.4f} GeV'.format(M_bb_2s))
```
